# Dataset Card: MCU (Marvel Cinematic Universe) Movies Dataset

## Ringkasan
Dataset ini berisi informasi mengenai film-film yang tergabung dalam **Marvel Cinematic Universe (MCU)**, mulai dari Phase 1 hingga Phase 6. Data mencakup informasi rating kritikus dan penonton, durasi film, serta performa finansial (budget produksi dan pendapatan box office) dari setiap judul film.

## Detail Dataset
- **Nama file**: `mcu.csv`
- **Asal file**: `https://www.kaggle.com/datasets/parthfr/marvel-cinematic-universe-boxoffice-csv`
- **Jumlah baris (records)**: 62 film
- **Jumlah kolom (features)**: 10 kolom
- **Format**: CSV (Comma-Separated Values)

## Deskripsi Kolom
| Kolom | Tipe Data | Deskripsi |
|---|---|---|
| `movie_title` | object (string) | Judul film MCU |
| `mcu_phase` | object (string) | Fase MCU dari film tersebut (Phase 1 - Phase 6) |
| `release_date` | object (string) | Tanggal rilis film |
| `tomato_meter` | int64 | Skor kritikus dari Rotten Tomatoes (Tomatometer, %) |
| `audience_score` | int64 | Skor penonton dari Rotten Tomatoes (%) |
| `movie_duration` | float64 | Durasi film (menit) |
| `production_budget` | float64 | Anggaran produksi film (USD) |
| `opening_weekend` | float64 | Pendapatan akhir pekan pembuka (USD) |
| `domestic_box_office` | float64 | Pendapatan box office domestik/Amerika Serikat (USD) |
| `worldwide_box_office` | float64 | Pendapatan box office di seluruh dunia (USD) |

## Potensi Kegunaan
Dataset ini cocok digunakan untuk analisis eksploratif (EDA), analisis tren performa box office tiap fase MCU, korelasi antara rating kritikus/penonton dengan pendapatan film, maupun sebagai dasar pemodelan prediktif pendapatan film.

## Catatan
Beberapa kolom finansial (`movie_duration`, `production_budget`, `opening_weekend`, `domestic_box_office`, `worldwide_box_office`) memiliki nilai yang hilang (missing values), kemungkinan karena data untuk film-film yang belum/baru dirilis belum tersedia secara lengkap.

## Visualizations

```python
df['worldwide_box_office'].hist(bins=10)
```

df['worldwide_box_office'].hist(bins=10) Distribusi worldwide_box_office memiliki nilai data yang condong bersifat 'Right-Skewed'.


*Data diambil dari Maximum - Minimum (2.799.439.100 - 206.136.825) = 2.593.302.275 dengan bins 10 maka lebar tiap bin = 259.330.227,5

Dari visual histogram terlihat bahwa sebagian besar data terkonsentrasi pada rentang pendapatan rendah hingga menengah (Bin 1 - Bin 3 dengan total 26 film dari 38 data yang memiliki nilai, atau sekitar 68,4%). Sedangkan film dengan pendapatan worldwide di atas 1,5 miliar USD jumlahnya sangat sedikit, sehingga menyebabkan ekor distribusi memanjang ke kanan (right-skewed).

```python
df['production_budget'].hist(bins=10)
```

df['production_budget'].hist(bins=10) Distribusi data 'production_budget' juga condong pada 'Right-Skewed' dikarenakan sebagian besar film memiliki anggaran produksi pada rentang 130 juta - 224 juta USD, sedangkan film dengan anggaran di atas 300 juta USD jumlahnya sangat sedikit, inilah penyebab data condong 'Right-Skewed'.

```python
df['production_budget'].plot.box()
```

Data anggaran produksi (production_budget) memiliki 50% nilainya dalam rentang 170.000.000 - 223.750.000 dengan nilai median sebesar 200.000.000. Dengan perhitungan IQR, kami mengasumsikan bahwa setiap nilai di atas 304.375.000 merupakan kandidat untuk outlier. Ini menunjukkan bahwa ada beberapa film dengan anggaran produksi yang jauh lebih besar dibandingkan mayoritas data yang ada.

Diketahui bahwa terdapat 38 data production_budget (24 data lainnya bernilai kosong/missing). Setelah data diurutkan dari nilai terkecil hingga terbesar, diperoleh nilai kuartil sebagai berikut:

Q1 (25%) = 170.000.000  
Q2 (50%) = Median = 200.000.000  
Q3 (75%) = 223.750.000  
IQR = Q3 - Q1 = 53.750.000  
Batas atas outlier = Q3 + 1.5 x IQR = 304.375.000

```python
df['worldwide_box_office'].plot.box()
```

Data pendapatan worldwide_box_office memiliki 50% nilai tengah pada rentang 486.881.376 - 1.147.955.776, dengan median sebesar 767.139.100. Nilai IQR diperoleh sebesar 661.074.399,75. Berdasarkan aturan Q3 + 1.5 x IQR, batas atas kandidat outlier adalah 2.139.567.375,6. Dengan demikian, pendapatan di atas nilai tersebut dapat dikategorikan sebagai kandidat outlier. Hal ini menunjukkan bahwa terdapat sejumlah kecil film dengan pendapatan yang jauh lebih tinggi dibandingkan mayoritas data pendapatan film lainnya (kemungkinan besar film-film blockbuster seperti Avengers: Endgame).

Disini kami menggunakan featured class / category 'mcu_phase' untuk melihat sebaran jumlah film di tiap fase MCU. Distribusi ini menunjukkan bahwa jumlah rilis film MCU cenderung meningkat pada fase-fase yang lebih baru (Phase 4 dan 5) dibandingkan fase-fase awal.

## Kesimpulan EDA

Berdasarkan proses data profiling dan visualisasi yang telah dilakukan terhadap dataset **MCU Movies**, dapat disimpulkan beberapa hal berikut:

1. **Ukuran dan kelengkapan data**: Dataset terdiri dari 62 film dengan 10 kolom. Kolom identitas dan rating (`movie_title`, `mcu_phase`, `release_date`, `tomato_meter`, `audience_score`) terisi lengkap, sedangkan kolom finansial (`movie_duration`, `production_budget`, `opening_weekend`, `domestic_box_office`, `worldwide_box_office`) memiliki 24 nilai kosong (missing values) dari 62 data, kemungkinan besar berasal dari film-film yang belum/baru dirilis sehingga data box office-nya belum tersedia. Tidak ditemukan data duplikat pada dataset ini.

2. **Distribusi pendapatan box office (`worldwide_box_office`)**: Bersifat **right-skewed**, artinya mayoritas film memiliki pendapatan menengah ke bawah (di bawah ~1 miliar USD), sementara hanya sedikit film yang mencapai pendapatan sangat tinggi (blockbuster seperti Avengers). Analisis boxplot memperkuat temuan ini, dengan batas atas outlier di sekitar 2,14 miliar USD, menandakan adanya beberapa film dengan pendapatan jauh di atas rata-rata.

3. **Distribusi anggaran produksi (`production_budget`)**: Juga bersifat **right-skewed**, dengan mayoritas film memiliki anggaran di rentang 130-224 juta USD, dan hanya sedikit film dengan anggaran di atas 300 juta USD yang menjadi kandidat outlier (batas atas ~304 juta USD).

4. **Distribusi film per fase (`mcu_phase`)**: Jumlah film tidak merata di setiap fase. Phase 5 memiliki jumlah film terbanyak (17 film / 27,4%), diikuti Phase 4 (15 film / 24,2%) dan Phase 3 (11 film / 17,7%), sedangkan Phase 1, Phase 2, dan Phase 6 memiliki jumlah film yang jauh lebih sedikit. Hal ini menunjukkan tren peningkatan jumlah rilis film MCU pada fase-fase yang lebih baru dibandingkan fase-fase awal.

5. **Implikasi**: Karena adanya *right-skewness* dan outlier pada variabel finansial, analisis lanjutan (misalnya pemodelan prediktif) sebaiknya mempertimbangkan transformasi data (seperti log-transform) atau penanganan outlier terlebih dahulu agar hasil analisis/model tidak bias oleh nilai-nilai ekstrem tersebut. Selain itu, missing values pada kolom finansial perlu ditangani (misalnya melalui imputasi atau pengecualian) sebelum melakukan analisis lebih lanjut yang melibatkan kolom-kolom tersebut.
